# DSGTask() Documentation

`DSGTask()`: An object for implementing a task in DSG.

**Object Variables**:
* `name`: name of task
* `best_val_loss`: best validation loss
* `best_loss_epoch`: epoch # of `best_val_loss`
* `dataset`: dataset for DSG task
* `converged`: boolean to see if task is converged
* `convergence_limit`: number of rounds of no improvements to consider as converged
* `diverged`: boolean to see if task is diverged
* `divergence_threshold`: threshold of loss to increase by to consider as diverged
* `final_round`: determine if task is in final stage of training and taken off DSG

**Object Methods**:
* `is_converged()`: helper function to return a boolean of whether or not the task is converged
* `is_diverged()`: helper function to return a boolean of whether or not the task is diverged
* `set_final_round()`: helper function to set final_round as true
* `should_keep_training()`: helper function to determine if task should continue to be trained on
* `update()`: helper function that should be called to update parameters of the object
