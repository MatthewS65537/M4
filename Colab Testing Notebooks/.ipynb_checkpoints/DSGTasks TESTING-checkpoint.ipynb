# DSGTasks() Documentation

`DSGTasks()`: An object for interacting with numerous `DSGTask()` objects

**Object Variables**:
* `tasks`: a list of `DSGTask()` objects
* `idx`: an integer index of the current object
* `num_tasks`: an integer representing the number of `DSGTask()` objects in `tasks`

**Object Methods**:
* `add_task()`: adds a `DSGTask()` object to the `tasks` list
* `reset_idx()`: resets the `idx` pointer
* `next()`: forwards the `idx` pointer and increments it by one
* `length()`: returns the number of tasks
* `all_converged()`: returns a boolean of whether or not all tasks are converged
* `should_keep_training()`: returns a boolean of whether or not training should continue (returns `True` as long as one task still needs training)

```python
dsg_tasks = DSGTasks()
```

```python
dsg_task1 = DSGTask("TEST1")
dsg_task2 = DSGTask("TEST2")
dsg_task3 = DSGTask("TEST3")
```

```python
dsg_tasks.add_task(dsg_task1)
dsg_tasks.add_task(dsg_task2)
dsg_tasks.add_task(dsg_task3)
```

```python
dsg_tasks.length()
```

```python
dsg_tasks.reset_idx()
for _ in range(dsg_tasks.length()):
    cur_idx = dsg_tasks.next()
    print(cur_idx)
```
